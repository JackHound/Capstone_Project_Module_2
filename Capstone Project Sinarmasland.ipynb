Outline ini dibuat untuk memahami dataset yang akan digunakan secara keseluruhan. Untuk mengetahui masing-masing komponen penyusun dataset, maka saya membuat sebuah function untuk melakukan pengecekan

```python
def data_quality_summary(df):
    summary = pd.DataFrame({
        "dtype": df.dtypes,
        "missing": df.isna().sum(),
        "missing_pct": df.isna().mean() * 100,
        "unique_values": df.nunique(),
        "duplicate_count": [
            df[col].dropna().duplicated().sum() if "_id" in col else pd.NA
            for col in df.columns
        ]
    })
    return summary
```

Pemakaian air negatif ditandai. Nilai tinggi diperiksa memakai batas Q3 + 3 * IQR di setiap cluster_category pada pemakaian non-negatif. Batas tersebut hanya alat penyaring statistik, bukan tarif atau batas teknis mater.

```python
b = billing_analysis
b['is_npi'] = b['payment_status'].isin(['Unpaid','Overdue'])
b['days_past_due'] = (Acuan - b['due_date']).dt.days.clip(lower=0)
b['overdue_asof'] = b['is_npi'] & b['due_date'].lt(Acuan)
b['pay_before_month'] = b['payment_date'].lt(b['billing_month'])
b['payment_after_asof'] = b['payment_date'].gt(Acuan)
b['before_handover_month'] = b['billing_month'].lt(b['handover_date'].dt.to_period('M').dt.to_timestamp())
b['negative_water'] = b['water_usage_m3'].lt(0)
b['amount_invalid'] = b['total_amount'].isna() | b['total_amount'].le(0)
b['below_fixed_fee'] = b['total_amount'].lt(b['ipl_base_fee'] + b['late_fee_penalty'])
```

## 4.6 Sengketa dan pembatalan penuh

Tanggal selesai diturunkan dari dispute_date + days_to_resolve dengan asumsi hari kalender. Status Dibatalkan baru menjadi dasar pengecualian ketika nilai koreksi sama dengan total_amount, tanggal selesai tidak melewati tanggal acuan, dan relasi ID cocok. Nilai koreksi tidak dikurangkan lagi setelah invoice dikeluarkan. Invoice Paid yang dibatalkan tetap dilaporkan sebagai kebutuhan rekonsiliasi refund/kredit; dataset tidak memiliki arus kas refund.

```python
b['resolution_date_est'] = b['dispute_date'] + pd.to_timedelta(b['days_to_resolve'],unit='D')
b['dispute_known_asof'] = b['dispute_id'].notna() & b['dispute_date'].le(Acuan)
b['cancel_recorded'] = b['resolution_status'].eq('Dibatalkan - Tagihan Salah')
b['cancel_confirmed'] = (b['cancel_recorded'] & b['dispute_known_asof']
    & b['resolution_date_est'].le(Acuan) & b['days_to_resolve'].ge(0)
    & np.isclose(b['amount_adjusted'],b['total_amount'],rtol=0,atol=.01))
b['dispute_pending_asof'] = b['dispute_known_asof'] & (
    b['resolution_status'].eq('Diproses') | b['resolution_date_est'].gt(Acuan))
dispute_status_table = pd.crosstab(b.loc[b['dispute_id'].notna(),'resolution_status'],
                                  b.loc[b['dispute_id'].notna(),'payment_status'])
display(dispute_status_table)
print('Pembatalan penuh memenuhi aturan:',int(b['cancel_confirmed'].sum()))
print('Pembatalan tercatat tetapi belum memenuhi aturan:',int((b['cancel_recorded'] & ~b['cancel_confirmed']).sum()))
print('Paid dengan pembatalan penuh:',int((b['cancel_confirmed'] & b['payment_status'].eq('Paid')).sum()))
```

## 6.1 Kontribusi township, cluster dan unit

Rasio NPI yang tinggi belum tentu berarti kontribusi terbesar. Tabel prioritas menunjukan jumlah NPI, rasio dan nominal bersama-sama. Peringkat kawasan merupakan titik awal penentuan beban, tindakan tetap mengikuti jalur kerja setiap invoice.

```python
def segment_summary(df, group):
    result = (df.groupby(group,dropna=False,observed=True)
        .agg(invoice=('invoice_id','size'),unit=('unit_id','nunique'),
             npi_invoice=('is_npi','sum'),nominal_npi_rp=('outstanding_recorded','sum')))
    result['npi_rate_pct'] = result['npi_invoice']/result['invoice']*100
    result['kontribusi_npi_pct'] = result['npi_invoice']/df['is_npi'].sum()*100
    return result.sort_values('npi_invoice',ascending=False)
```

## 6.7 Pemeriksaan statistik: perbedaan di dalam cluster

Satu unit mempunyai banyak invoice. Jika seluruh invoice dianggap pengamatan independen, ketidakpastian dapat terlihat terlalu kecil. Pemeriksaan berikut lebih dulu menghitung bagian NPI per unit, kemudian membandingkan rata-ratanya untuk dua kelompok di dalam cluster yang memiliki kedua kelompok.

Selisih tiap cluster dirata-ratakan dengan bobot jumlah unit cluster yang sama untuk kedua kelompok. Bootstrap mengambil ulang **cluster**, bukan baris invoice, sebanyak 2.000 kali. Interval persentil 95% digunakan sebagai pemeriksaan kestabilan antarkawasan. Ini bukan bukti sebab-akibat, bukan interval prediksi dua kuartal, dan bukan ketidakpastian pencacahan dataset lengkap. Karakteristik unit dan waktu perubahan atribut yang tidak tercatat masih dapat memengaruhi hasil.


```python
def within_cluster_bootstrap(unit_df, feature, label, iterations=2000):
    temp = unit_df.groupby(['cluster_id',feature],observed=True).agg(
        mean_npi=('npi_share_pct','mean'),unit=('npi_share_pct','size'))
    means = temp['mean_npi'].unstack(feature)
    counts = temp['unit'].unstack(feature)
    common = means.dropna(subset=[0,1]).index
    means,counts = means.loc[common],counts.loc[common]
    difference = (means[1]-means[0]).to_numpy()
    weights = counts[[0,1]].sum(axis=1).to_numpy()
    if len(common)<2: raise ValueError('Cluster pembanding tidak mencukupi.')
    estimate = np.average(difference,weights=weights)
    rng = np.random.default_rng(42)
    draw = rng.integers(0,len(common),size=(iterations,len(common)))
    distribution = ((difference[draw]*weights[draw]).sum(axis=1)/weights[draw].sum(axis=1))
    low,high = np.quantile(distribution,[.025,.975])
    return {'perbandingan':label,'cluster_dengan_kedua_kelompok':len(common),
            'unit_diperhitungkan':int(weights.sum()),
            'rata_kelompok_0_pct':np.average(means[0],weights=weights),
            'rata_kelompok_1_pct':np.average(means[1],weights=weights),
            'selisih_1_minus_0_pp':estimate,'interval_95_bawah_pp':low,'interval_95_atas_pp':high}

unit_stats = unit_summary.copy()
unit_stats['contact_missing'] = unit_stats['contact_missing'].astype(int)
bootstrap_results = pd.DataFrame([
    within_cluster_bootstrap(unit_stats,'is_vacant','Kosong (1) vs tidak kosong (0)'),
    within_cluster_bootstrap(unit_stats,'contact_missing','Kontak kosong (1) vs tersedia (0)')])
display(bootstrap_results)
```

Setelah membandingkan di dalam 111 cluster, selisih rata-rata bagian NPI unit kosong tetap sekitar 29,45 poin persentase, dengan interval bootstrap 28,91–29,99. Untuk kontak kosong dibandingkan semua kontak tersedia, selisihnya 9,85 poin persentase, dengan interval 9,15–10,59. Pola positif tetap terlihat sesudah komposisi cluster disamakan; kedua temuan layak menjadi dasar pemilihan kelompok pilot.

Rata-rata pada tabel statistik memberikan bobot setara untuk unit di dalam masing-masing kelompok dan bobot cluster yang sama antarkelompok. Karena itu, angkanya tidak harus sama dengan NPI portofolio berbasis invoice pada Bab 5. Pembanding kontak di sini juga menggabungkan kontak terverifikasi dan belum terverifikasi. ## 7.1 Target yang dapat dibandingkan

Saya mengusulkan penurunan **2 poin persentase** dari B1 dalam enam bulan sebagai target awal untuk diuji bersama kapasitas tim. Target ini belum merupakan komitmen atau hasil optimasi; biaya tindakan, kapasitas agen, dan tingkat keberhasilan pelunasan terverifikasi belum tersedia.

Evaluasi memakai 790.817 invoice B1 yang dibekukan sebagai kelompok awal. Pelunasan penuh mengurangi pembilang NPI, sementara penyebut tetap. Kasus pembatalan, penghapusan, dan koreksi berikutnya masuk catatan rekonsiliasi terpisah. Pembayaran sebagian dicatat sebagai kas masuk, tetapi belum dianggap menyembuhkan satu invoice untuk KPI pelunasan penuh ini.

Enam bulan dibagi menjadi Februari–April dan Mei–Juli 2025. Pembagian ini merupakan dua periode tiga bulan yang berurutan, **bukan kuartal kalender**. Karena data sesudah Januari 2025 tidak tersedia, bagian ini sepenuhnya merupakan simulasi historis.


```python
base_denominator = len(active)
base_npi = int(active.is_npi.sum())
base_rate = base_npi/base_denominator*100
target_reduction_pp = 2.0
target_cures = math.ceil(base_denominator*target_reduction_pp/100)
eligible = active.loc[active.work_route.eq('Collection: kontak tersedia dan terverifikasi')].copy()
eligible_n = len(eligible)
eligible_amount = eligible.total_amount.sum()
target_pool_cure_pct = target_cures/eligible_n*100
equal_stage_cure = 1-np.sqrt(1-target_cures/eligible_n)
scenario_pool = pd.DataFrame([{
    'invoice_awal_B1':base_denominator,'NPI_awal':base_npi,'NPI_awal_pct':base_rate,
    'invoice_pool_collection':eligible_n,'unit_pool_collection':eligible.unit_id.nunique(),
    'nominal_pool_rp':eligible_amount,'target_lunas_penuh':target_cures,
    'target_NPI_maksimum_pct':base_rate-target_reduction_pp,
    'pelunasan_pool_diperlukan_pct':target_pool_cure_pct,
    'pelunasan_per_tahap_setara_pct':equal_stage_cure*100}])
display(scenario_pool.T.rename(columns={0:'nilai'}))
print('Kebutuhan rata-rata pelunasan per minggu (asumsi 26 minggu):',math.ceil(target_cures/26),'invoice.')
```

## 7.2 Sensitivitas tingkat pelunasan

Asumsi 5%, 10%, dan 15% per tahap dipakai untuk melihat konsekuensi perubahan tingkat pelunasan, bukan sebagai estimasi dari respons Lunas Setelah Kontak. Pada tahap kedua, persentase diterapkan hanya pada sisa invoice dalam pool yang belum lunas pada tahap pertama. Dengan begitu, invoice yang sama tidak dihitung dua kali.

Ilustrasi nominal memakai rata-rata nominal pool yang sama untuk setiap invoice yang lunas. Pemilihan invoice nyata dapat menghasilkan nominal berbeda. Angka ini merupakan potensi penerimaan bruto bersyarat, bukan proyeksi kas bersih; biaya, pengembalian, dan pembayaran parsial belum tersedia. Invoice yang baru lolos rekonsiliasi/perbaikan kontak tidak ditambahkan ke pool simulasi.


```python
rows = []
for label,rate in [('0%: pembanding aritmetis',0.0),('5% per tahap',.05),
                   ('10% per tahap',.10),('15% per tahap',.15)]:
    cured_1 = math.floor(eligible_n*rate)
    cured_2 = math.floor((eligible_n-cured_1)*rate)
    cured_total = cured_1+cured_2
    rows.append({'skenario':label,'pelunasan_per_tahap_pct':rate*100,
        'lunas_tahap_1':cured_1,'lunas_tahap_2_tambahan':cured_2,'lunas_kumulatif':cured_total,
        'npi_akhir_tahap_1_pct':(base_npi-cured_1)/base_denominator*100,
        'npi_akhir_tahap_2_pct':(base_npi-cured_total)/base_denominator*100,
        'turun_pp':cured_total/base_denominator*100,
        'ilustrasi_penerimaan_bruto_rp':cured_total/eligible_n*eligible_amount,
        'target_2pp_tercapai':cured_total>=target_cures})
scenario_results = pd.DataFrame(rows)
display(scenario_results)
assert scenario_results.lunas_kumulatif.le(eligible_n).all()
assert scenario_results.lunas_kumulatif.eq(
    scenario_results.lunas_tahap_1+scenario_results.lunas_tahap_2_tambahan).all()
assert scenario_results.npi_akhir_tahap_2_pct.ge(0).all()
fig,ax = plt.subplots(figsize=(10,4.5))
for row,color in zip(rows,['#7A7A7A','#C77349','#799C9C','#245C5B']):
    ax.plot([0,3,6],[base_rate,row['npi_akhir_tahap_1_pct'],row['npi_akhir_tahap_2_pct']],
            marker='o',label=row['skenario'],color=color)
ax.axhline(base_rate-target_reduction_pp,linestyle='--',color='#777777',label='Usulan target: turun 2 pp')
ax.set(title='Simulasi pelunasan B1 — bukan realisasi / ramalan',
       xlabel='Bulan setelah tanggal acuan',ylabel='NPI pada kelompok awal (%)',xticks=[0,3,6],ylim=(0,40))
ax.legend(loc='lower left',fontsize=8)
fig.tight_layout(); plt.show()
```

Pada asumsi 5% per tahap, terdapat 15.666 pelunasan dan NPI akhir 34,69%; target 2 poin persentase belum tercapai karena masih kurang 151 invoice. Pada asumsi 10%, NPI menjadi 32,81%; pada 15%, menjadi 31,03%. Angka tersebut hanya menunjukkan hubungan aritmetis antara tingkat pelunasan yang diasumsikan dan hasil KPI.
